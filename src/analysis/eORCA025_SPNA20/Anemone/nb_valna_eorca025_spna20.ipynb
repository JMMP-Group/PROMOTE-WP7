```python
import matplotlib.pyplot as plt
import xarray as xr
```

```python
# -- eORCA025 SPNA20 Global Parent - PROMOTE -- #
ds_dy838o = xr.open_mfdataset("/dssgfs01/scratch/otooth/PROMOTE/PROMOTE-WP7/nemo_pipeline/work/eORCA025_SPNA20/valna/diagnostics/eORCA025_SPNA20_VALNA_*.nc")

ds_dy838o
```

### Gulf Stream Separation Latitude:

```python
# -- Figure: Gulf Stream Separation Latitude -- #
plt.figure(figsize=(8, 4))
plt.grid(visible=True, linewidth=1, linestyle="--", color="0.3", alpha=0.1)

# eORCA025 - GOSI10 [dm2570o]:
# ds_dm257o["gs_latitude"].rolling(time_counter=24, center=True).mean().plot(linewidth=3, color="0.1", label="eORCA025 GOSI10")
# eORCA025 SPNA20 - PROMOTE [dy838o]:
ds_dy838o["gs_latitude"].rolling(time_counter=24, center=True).mean().plot(linewidth=3, color="coral", label="eORCA025 SPNA20")

plt.title("Gulf Stream Separation Latitude", fontsize=13, fontweight="bold")
plt.xlabel("Time", fontsize=12, fontweight="bold")
plt.ylabel("Latitude [°N]", fontsize=12, fontweight="bold")

plt.legend(fontsize=10, loc="best")
```
